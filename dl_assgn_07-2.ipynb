# Assignment 7: Implementing Loss Functions (MSE and Cross-Entropy)

## Step 1: Import the Required Library

```python
import numpy as np
```

## Step 2: Set Up the True Labels and Predicted Probabilities

```python
# Actual class labels
y_actual = np.array([1, 0, 1, 1, 0])

# Predicted probabilities
y_predicted = np.array([0.9, 0.2, 0.8, 0.6, 0.1])

print("Actual Values:   ", y_actual)
print("Predicted Values:", y_predicted)
```

## Step 3: Compute the Mean Squared Error (MSE)

```python
mse = np.mean((y_actual - y_predicted) ** 2)

print(f"Mean Squared Error: {mse:.4f}")
```

## Step 4: Compute the Cross-Entropy Loss

```python
# Avoid log(0)
epsilon = 1e-15

y_predicted_clipped = np.clip(y_predicted, epsilon, 1 - epsilon)

cross_entropy = -np.mean(
    y_actual * np.log(y_predicted_clipped) +
    (1 - y_actual) * np.log(1 - y_predicted_clipped)
)

print("Cross-Entropy Loss:", cross_entropy)
```

## Step 5: Side-by-Side Comparison of Both Losses

```python
print("Loss Function Comparison")
print("------------------------")
print(f"Mean Squared Error : {mse:.4f}")
print(f"Cross-Entropy Loss : {cross_entropy:.4f}")
```

## Conclusion

MSE is better suited to regression tasks, while cross-entropy loss is the standard choice for classification, since it penalizes confident wrong predictions much more heavily.
